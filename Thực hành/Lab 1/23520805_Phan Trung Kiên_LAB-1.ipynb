#### Câu 34: Nhập vào số nguyên tố dương n. Kiểm tra số nguyên tố

```python
num = int(input("Nhập vào số nguyên tố n: "))
is_Prime = True

if num < 2:
    is_Prime = False
    print(f"{num} không phải số nguyên tố")
else:
    for i in range(2, int(num ** 0.5) + 1): # Chạy từ 2 đến căn bậc 2 của num
        if num % i == 0:
            is_Prime = False
            break
    if is_Prime == True :
        print(f"{num} là số nguyên tố")
    else:
        print(f"{num} không phải là số nguyên tố")
```

#### Câu 46: Liệt kê tất cả nghiệm nguyên dương của phương trình: 2x+7y+9z = 979 (x,y,z > 0)

```python
for z in range(1, 979//9 + 1): #z thuộc [1, 108]
    S = 979 - 9*z

    for y in range(1, (S//7) + 1): #7y <= S => y <= S//7
        if (S - 7*y) % 2 == 0: #Kiểm tra x nguyên
            x = (S - 7*y) // 2
            if x > 0: #Kiểm tra x dương
                print(f"x = {x}, y = {y}, z = {z}")
```

#### Câu 50: Trò chơi "Dãy số may mắn"

```python
import random

n = int(input("Nhập số vé: "))
ve = []

for i in range(n):
    nums = list(map(int, input(f"Nhập vé {i+1}: ").split("-")))
    ve.append(nums)

ketqua = random.sample(range(1, 46),6)
print("Kết quả: ", ketqua)

tong_tien = 0

for v in ve:
    trung = len(set(v) & set(ketqua))

    if trung == 3: tong_tien += 30000
    elif trung == 4: tong_tien += 300000
    elif trung == 5: tong_tien += 10000000
    elif trung == 6: tong_tien += 10000000000

print("Tổng tiền người chơi trúng: ", tong_tien)
print("Tổng tiền khi đã trừ tiền mua vé: ", tong_tien - (10000 * n))
```
