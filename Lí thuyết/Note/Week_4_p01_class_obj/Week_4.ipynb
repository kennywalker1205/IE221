## _Bài 5_: **Function** (Hàm Lambda)

##### VD: y = 2x + 1

```python
# Dùng hàm thông thường
def ham(x):
    return 2*x + 1
```

```python
ham(6)
```

```python
# Dùng hàm lambda
(lambda x: 2*x + 1 + ham(3))(6) #Có thể lồng hàm bình thường
```

```python
# VD:
y = (lambda x: print(2*x + 1))(6)
    # Khi in giá trị
print(y)
```

```python
# Vận dụng
ds = [6, 2, -8, 5, 9, 3]

a = sorted(ds, key = lambda x: -x)
print(a)
b = max(filter(lambda x: x > 0, ds))
print(b)
c = max(filter(lambda x: x < 0, ds))
print(c)
d = min(filter(lambda x: x > 0, ds))
print(d)
```
