## Nhóm 1: Toán tử số học

```python
# Phép cộng (+)
print("Số + số:", 10 + 5, ", ", 10.5 + 5.5)
print("Số + bool:", True + 5, ", ", False + 2)

print("Str + Str:", "Hello" + " " + "World")

print("List + List:", [1, 2] + [3, 4])

print("Tuple + Tuple:", (1, 2, 3) + (4, 5))
```

```python
# Phép trừ (-)
print("Số - số:", 10 - 5, ", ", 10.5 - 5.5, ", ", -10 - 5)
print("Số - bool:", True - 5, ", ", 2 - False)

print("Set - Set:", {1, 2, 3} - {2})
```

```python
# Phép nhân (*)
print("Số * Số:", 5 * 3)
print("Số * Str:", 5 * "ab")
print("Số * list:", [1, 2] * 2, " ", [1, 2] * -1)
print("Số * Tuple:", (1, 2, 3) * 4)
print("Số * bool:", True * 5, ", ", False * 2)

print("Str * bool:", "ab" * False, "cd" * True)

print("List * bool:", [1, 2] * True, ", ", [3, 4] * False)

print("Tuple * bool:", (1, 2, 3) * True, ", ", [3, 4] * False)

print("Bool * bool:", True * True, ", ", True * False)
```

```python
# Phép chia (/)
print("Số / Số:", 10 / 2, ", ", 10 / 3)
print("Số / bool:", True / 2, ", ", False / 3, ", ", 10 / True)

print("Bool / bool:", True / True, ", ", False / True)
```

```python
# Phép chia phần nguyên (//)
print("Số // Số:", 10 // 2, ", ", 10 // 3)
print("Số // bool:", True // 1, ", ", False // 3)

print("Bool // bool:", True // True, ", ", False // True)
```

```python
# Phép chia lấy phần dư (%)
print("Số % Số", 10 % 2, ", ", 10 % 3)
print("Số % bool:", True % 2, ", ", False % 3)

print("Bool % bool:", True % True, ", ", False % True)
```

```python
# Phép lũy thừa (**)
print("Số ** Số:", 2 ** 3, ", ", 2 ** -3, ", ", 2 ** 0.5)
print("Số ** bool:", 2 ** True, ", ", 2 ** False)
```

## Nhóm 2: Toán tử so sánh


```python
# Số
print("Số với Số:",
        10 > 5,
        10 >= 10,
        10 < 9,
        10 <= 11,
        10 == 1,
        10 != 1 )

print("Số với Bool:",
        10 > True,
        -10 >= True,
        10 < False,
        -10 <= True,
        10 == True,
        10 != True )

print("Số với Str/List/Tuple/Dict:",
        10 == "ab", 10 == '10',
        10 != "cd", 10 != '10',

        10 == [10], 10 != [10],

        10 == (10), 10 != (10),

        10 == {'ab': 10}, 10 != {'ab': 10},
      )
```

```python
# Chuỗi (Str)
print("Str với Str:",
      "ab" == "ab" , "ab" != "cd",
      "ab" > "cd" , "ab" >= "cd",
      "ab" < "cd" , "ab" <= "cd",
      )

print("Str với List/Tuple/Dict/Set/Bool:",
      "ab" == [1, 2], "ab" != [1, 2],
      "ab" == ["a", "b"], "ab" != ["a", "b"],

      "ab" == (1, 2), "ab" != (1, 2),
      "ab" == ("a", "b"), "ab" != ("a", "b"),

      "a" == {"a" : 1}, "a" != {"a": 1},

      "ab" == {1, 2}, "ab" != {1, 2},
      "ab" == {"a", "b"}, "ab" != {"a", "b"},

      "True" == True, "False" != False
      )
```

```python
#List
list_a = [1, 2, 3]
list_b = [1, 2, 5]

print("List với List:",
    list_a > list_b, list_a >= list_b,
    list_a < list_b, list_a <= list_b,
    list_a == list_b, list_a != list_b,
      )
print("List với Tuple/Dict/Set/Bool:",
    list_a == (1, 2, 3), list_a != (1, 2, 3),

    list_a == [1, 2, 3], list_a != [1, 2, 3],

    list_a == {"a": 1, "b": 2}, list_a != {"a": 1, "b": 2},

    list_a == {1, 2, 3}, list_a != {1, 2, 3},

    list_a == "True", list_a != "False"
      )
```

```python
#Tuple
tuple_a = (1, 2, 3)
tuple_b = (1, 2, 4)

print("Tuple với Tuple:",
      tuple_a > tuple_b, tuple_a >= tuple_b,
      tuple_a < tuple_b, tuple_a <= tuple_b,
      tuple_a == tuple_b, tuple_a != tuple_b,
      )

print("Tuple với Dict/Set/Bool:",
      tuple_a == {1: 2}, tuple_a != {1: 2},

      tuple_a == {1, 2, 3}, tuple_a != {1, 2, 3},

      tuple_a == "True", tuple_a != "False"
      )
```

```python
# Dict
dict_1 = {"a": 1, "b": 2}
dict_2 = {"b": 2, "a": 1}

print("Dict với Dict/Set/Bool:",
      dict_1 == dict_2, dict_1 != dict_2,
      dict_1 == {1: 2}, dict_2 != {1: 2},

      dict_1 == {1, 2}, dict_1 != {2, 1},

      dict_1 == True, {} != True,
      )
```

```python
# Bool
print("Bool với Bool:",
      True == False, True != False,
      True > False, True < False,
      True >= False, True <= False,
    )
```

## Nhóm 3. Toán tử Logic

```python
# Số
print("Toán tử Logic với số:",
      5 and 3, 5 or 3, not 3,
      #Số với chuỗi
      5 and "ab", 5 or "ab",
      0 and "ab", 0 or "ab",
      #Số với list
      5 and [1, 2, 3], 5 or [1, 2, 3],
      #Số với tuple
      5 and (1, 2, 3), 5 or (1, 2, 3),
      #Số với dict
      5 and {'a': 1}, 5 or {'a': 1},
      #Số với set
      5 and {1, 2, 3}, 5 or {1, 2, 3},
      #Số với bool
      5 and True, 5 or True
)
```

```python
# Str
print("Toán tử Logic với Str:",
      # Str với Str
      "ab" and "cd", "ab" or "cd", not "ab",

      # Str với List
      "\nStr với List:", "ab" and [1, 2], "ab" or [1, 2],

      # Str với Tuple
      "\nStr với Tuple:", "ab" and (1, 2), "ab" or (1, 2),

      # Str với Dict
      "\nStr với Dict:", "ab" and {'k': 1}, "ab" or {'k': 1},

      # Str với Set
      "\nStr với Set:", "ab" and {1, 2}, "ab" or {1, 2},

      # Str với Bool
      "\nStr với Bool:", "ab" and True, "ab" or True
)
```

```python
print("Toán tử Logic với List:",
      [1, 2] and [3, 4], [1, 2] or [3, 4], not [1, 2],

      # List với Số
      "\nList với Số:", [1, 2] and 5, [1, 2] or 5,

      # List với Str
      "\nList với Str:", [1, 2] and "ab", [1, 2] or "ab",

      # List với Tuple
      "\nList với Tuple:", [1, 2] and (1, 2), [1, 2] or (1, 2),

      # List với Dict
      "\nList với Dict:", [1, 2] and {'k': 1}, [1, 2] or {'k': 1},

      # List với Set
      "\nList với Set:", [1, 2] and {1, 2}, [1, 2] or {1, 2},

      # List với Bool
      "\nList với Bool:", [1, 2] and True, [1, 2] or True
)
```

```python
print("Toán tử Logic với Dict:",
      {'a': 1} and {'b': 2}, {'a': 1} or {'b': 2}, not {'a': 1},

      # Dict với Số
      "\nDict với Số:", {'a': 1} and 5, {'a': 1} or 5,

      # Dict với Str
      "\nDict với Str:", {'a': 1} and "ab", {'a': 1} or "ab",

      # Dict với List
      "\nDict với List:", {'a': 1} and [1, 2], {'a': 1} or [1, 2],

      # Dict với Tuple
      "\nDict với Tuple:", {'a': 1} and (1, 2), {'a': 1} or (1, 2),

      # Dict với Set
      "\nDict với Set:", {'a': 1} and {1, 2}, {'a': 1} or {1, 2},

      # Dict với Bool
      "\nDict với Bool:", {'a': 1} and True, {'a': 1} or True
)
```

```python
print("Toán tử Logic với Set:",
      {1, 2} and {3, 4}, {1, 2} or {3, 4}, not {1, 2},

      # Set với Số
      "\nSet với Số:", {1, 2} and 5, {1, 2} or 5,

      # Set với Str
      "\nSet với Str:", {1, 2} and "ab", {1, 2} or "ab",

      # Set với List
      "\nSet với List:", {1, 2} and [1, 2], {1, 2} or [1, 2],

      # Set với Tuple
      "\nSet với Tuple:", {1, 2} and (1, 2), {1, 2} or (1, 2),

      # Set với Dict
      "\nSet với Dict:", {1, 2} and {'k': 1}, {1, 2} or {'k': 1},

      # Set với Bool
      "\nSet với Bool:", {1, 2} and True, {1, 2} or True
)
```

```python
print("Toán tử Logic với Bool:",
      True and False, True or False, not True,

      # Bool với Số
      "\nBool với Số:", True and 5, True or 5,

      # Bool với Str
      "\nBool với Str:", True and "ab", True or "ab",

      # Bool với List
      "\nBool với List:", True and [1, 2], True or [1, 2],

      # Bool với Tuple
      "\nBool với Tuple:", True and (1, 2), True or (1, 2),

      # Bool với Dict
      "\nBool với Dict:", True and {'k': 1}, True or {'k': 1},

      # Bool với Set
      "\nBool với Set:", True and {1, 2}, True or {1, 2}
)
```
